# Creating Working Copies

```python
matches_clean = matches.copy()
deliveries_clean = deliveries.copy()
```

# Team Name Standardization

```python
unique_team = pd.unique(matches_clean[['team1','team2','winner']].values.ravel())
unique_team
```

```python
team_map = {
    'Royal Challengers Bangalore' : 'Royal Challengers Bengaluru',
    'Kings XI Punjab' : 'Punjab Kings',
    'Delhi Daredevils' : 'Delhi Capitals',
    'Rising Pune Supergiants' : 'Rising Pune Supergiant'
}
```

```python
matches_clean = matches_clean.replace(team_map)
deliveries_clean = deliveries_clean.replace(team_map)
```

```python
matches_clean['team1'].unique()
```

# Export Cleaned Dataset

```python
matches_clean.to_csv('../data/cleaned/matches_clean.csv', index=False)
deliveries_clean.to_csv('../data/cleaned/deliveries_clean.csv', index=False)
```
